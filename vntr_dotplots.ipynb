```python
import matplotlib.pyplot as plt
import matplotlib
from pysam import AlignmentFile
import numpy as np
import pandas as pd
from glob import glob
import os

from lib.visualize import plot_breakpoints, plot_dotplots
from lib.utils import compute_aln_matrix

plt.style.use('ggplot')
```

### Assemblies: Chromothripsis 

```python
bam_name = '/confidential/tGenVar/vntr/output_maryam/pipeline/results/hg38/asm_align/merged_chromothripsis.bam'

for i in range(len(vntr_regions)):
    chrom = vntr_regions.iloc[i]['chrom']
    start = vntr_regions.iloc[i]['start']
    stop = vntr_regions.iloc[i]['stop']
    filename = 'vntr_dotplots/asmb_chromo/asmb_chromo_' + chrom + ':' + str(start) + '-' + str(stop) + '.png'
    plot_dotplots(bam_name, chrom, start, stop, padding=500, color_by='', outfile=filename)
```
